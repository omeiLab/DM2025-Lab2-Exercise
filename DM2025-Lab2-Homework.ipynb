## 1. Model Development (10 pts Required)

### 1.1 Preprocessing Steps

The objective of this section is to process our data from `json` fromat to tabular data. This can be easily done by loading it to a `pd.DataFrame` and then using `pd.json_normalize()` function to flatten the nested JSON data. 

Other processing includes:
- Identify and split the train and test data sets.

- Add the emotion labels to the training data.
- Concatenate the hashtags by whitespace if there are multiple hashtags in a single post.
- Rename the columns to meaningful and easy to understand names.
- Remove the feature `type` since there's only 1 value for all rows.

### 1.2 Feature Engineering Steps

Here we perform multiple feature engineering on text:

- `replace_mentions`: Replace `@<user>` mentions with a special token `[MENTION]`.

- `remove_tags`: Remove tags in content such as `#<hashtag>`.
- `replace_ellipsis_with_token`: Replace `...` with a special token `[ELIP]`.
- `add_space_brackets`: Add space before and after brackets such as `[ ]`. This can help the model to identify the brackets as separate special tokens.
- `remove_escape_chars`: Remove `\n`, `\t` and `\r` escape characters.
- `remove_uncommon_punctuation`: Remove some uncommon punctuation marks.
- `remove_suffix_comma`: I notice that there are some posts with a comma at the end, which is not a common practice. So, I removed it.
- `remove_quote`: I notice that there are some posts with quotes, I remove them in order to keep the formats consistent.
- `clean_non_english_but_keep_emoji`: Remove non-ASCII characters but keep emojis.

Finally, we convert all the letters to lowercase and remove extra spaces.

Later we will use RoBERTa for text classification. So here we concatenate the text and the hashtags to input to RoBERTa.

$$[CLS] \quad \text{text} \quad [SEP] \quad \text{hashtags}$$

For the emotion labels, we use a simple label encoding by mapping them to 0 to 5. 

### 1.3 Explanation of Your Model

We use the pretrained RoBERTa model [SamLowe/roberta-base-go_emotions](https://huggingface.co/SamLowe/roberta-base-go_emotions) from Hugging Face's model hub for text classification. We fine-tune the model on our dataset and use it to predict the emotion labels for the test data. Note that we have some additional special tokens such as `[TAG]` for model to learn, so we have to add them to the RoBERTa tokenizer and resize the embedding layer accordingly.

To address the imbalancing issue of the dataset, we apply class weighting to the loss function during training. The class weighting is calculated by $$w_i=\frac1{\sqrt n_i}$$ where $n_i$ is the number of samples in class $i$. 

We fine-tune the model on our dataset for 2 stages:
- 15 epochs with a peak learning rate of 5e-6 controlled by `get_linear_schedule_with_warmup`.

- 5 epochs with a stable learning rate of 1e-5.

The evaluation metric used is macro F1-score, which is the average of the F1-scores of each class.

## 1. Preprocessing Steps

````python
import pandas as pd
import json

# Load JSON data from a file
with open('compt/final_posts.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

# Convert JSON --> DataFrame
df = pd.DataFrame(data) 

# For nested JSON, use json_normalize to flatten it
df = pd.json_normalize(data) 

df.head()
````

````python
# load other csv
identification = pd.read_csv('compt/data_identification.csv')
emotion = pd.read_csv('compt/emotion.csv')
````

````python
# identify train & test data
df['for_train'] = df['root._source.post.post_id'].isin(identification[identification['split'] == 'train']['id'])

# convert hashtag, list --> string
# [] --> "" , ["a", "b"] --> "a b"
df['hashtag'] = df['root._source.post.hashtags'].apply(lambda x: ' '.join(x) if isinstance(x, list) else '')

# rename columns for easier access
df = df.rename(columns={
	'root._source.post.post_id': 'id',
	'root._source.post.text': 'text',
})

# choose relevant columns only
# root._type is removed since there is only 1 type: "post"
df = df[['id', 'text', 'hashtag', 'for_train']]

df.head()
````

````python
# split train & test data
train_df = df[df['for_train'] == True].reset_index(drop=True)
test_df = df[df['for_train'] == False].reset_index(drop=True)

# concat emotion labels to train data
train_df = pd.merge(train_df, emotion, on='id', how='left')

# drop 'for_train' column
train_df = train_df.drop(columns=['for_train'])
test_df = test_df.drop(columns=['for_train'])
````

## 2. Feature Engineering Steps

````python
import re

# replace @<user> --> [MENTION]
def replace_mentions(text):
    return re.sub(r'(\B@\w+)', '[MENTION]', text)

# remove tags
def remove_tags(text):
    return re.sub(r'(\B#\w+)', '', text)

# add space before and after []
def add_space_brackets(text):
	text = re.sub(r'\[', ' [', text)
	text = re.sub(r'\]', '] ', text)
	return text

# example
sample_1 = "@noveliciouss OMG, that's such a beautiful quote[NAME]. #breathless #amwriting"
print(f"Result: {add_space_brackets(replace_tags(replace_mentions(sample_1)))}")
````

````python
# remove escape characters: \n, \t, \r
def remove_escape_chars(text):
    text = text.replace('\n', ' ')
    text = text.replace('\t', ' ')
    text = text.replace('\r', ' ')
    return text

# remove suffix ,
def remove_suffix_comma(text):
	return text[:-1] if text.endswith(',') else text

# remove prefix " " and suffix " "
def remove_quote(text):
	return text[1:-1] if text.startswith('"') and text.endswith('"') else text

# remove extra spaces
def remove_extra_spaces(text):
	return re.sub(r'\s+', ' ', text).strip()

# remove unmeaningful or uncommon punctuation: *, &, %, $, ^, ~, `, |, \
def remove_uncommon_punctuation(text):
	return re.sub(r'[\*\&\%\$\^\~\`\|\\]', '', text)

# assign ... a special token [ELIP]
def replace_ellipsis_with_token(text):
    text = re.sub(r'\.{2,}', ' [ELIP] ', text)
    return text

# example
sample_2 = "This is a test...\nNew line here!\tAnd a   tab. $$$"
print(f"Result: {remove_extra_spaces(replace_ellipsis_with_token(remove_uncommon_punctuation(remove_escape_chars(sample_2))))}")
````

````python
import emoji

# remove non-ascii characters while keeping emojis
def clean_non_english_but_keep_emoji(text):
    temp_text = emoji.demojize(text, delimiters=(" EMOJI_START ", " EMOJI_END "))
    cleaned_text = re.sub(
        r"[^a-zA-Z0-9\s.,!?:;'\-()\[\]{} EMOJI_START EMOJI_END]", 
        "", 
        temp_text,
        flags=re.UNICODE  
    )
    final_text = emoji.emojize(cleaned_text, delimiters=(" EMOJI_START ", " EMOJI_END "))
    final_text = re.sub(r'\s+', ' ', final_text).strip()
    return final_text

# example
sample_3 = "Thanks for this деньги 😂,"
print(f"Result: {clean_non_english_but_keep_emoji(sample_3)}")
````

````python
# combine all the data cleaning functions
def data_cleaning(text):
	text = replace_mentions(text)
	text = remove_tags(text)
	text = replace_ellipsis_with_token(text)
	text = add_space_brackets(text)
	text = remove_escape_chars(text)
	text = remove_uncommon_punctuation(text)
	text = remove_suffix_comma(text)
	text = remove_quote(text)
	text = clean_non_english_but_keep_emoji(text)
	text = remove_extra_spaces(text)
	text = text.lower().strip()
	return text

# apply on train & test data
train_df['cleaned_text'] = train_df['text'].apply(data_cleaning)
train_df['cleaned_hashtag'] = train_df['hashtag'].apply(data_cleaning)
test_df['cleaned_text'] = test_df['text'].apply(data_cleaning)
test_df['cleaned_hashtag'] = test_df['hashtag'].apply(data_cleaning)

train_df.head()
````

````python
# encode emotion labels
emotion_mapping = {
	'anger': 0,
    'disgust': 1,
	'fear': 2,
	'sadness': 3,
	'surprise': 4,
    'joy': 5
}

train_df['label'] = train_df['emotion'].map(emotion_mapping)

train_df.head()
````

````python
# concat cleaned_text & cleaned_hashtag by [SEP]
def combine_text_and_tags(row):
    text = row['cleaned_text']
    tags = row['cleaned_hashtag']

    if tags and tags.strip():
        return f"{text} [SEP] {tags}"
    else:
        return text

train_df['final_input_text'] = train_df.apply(combine_text_and_tags, axis=1)
test_df['final_input_text'] = test_df.apply(combine_text_and_tags, axis=1)

# filter columns
train_df = train_df[['id', 'final_input_text', 'label']]
test_df = test_df[['id', 'final_input_text']]

# save to csv
train_df.to_csv('compt/train_data.csv', index=False)
test_df.to_csv('compt/test_data.csv', index=False)
````

## 3. Model Implementation Steps

````python
from transformers import AutoTokenizer, AutoModel
from transformers.data.data_collator import DataCollatorWithPadding
from transformers import get_linear_schedule_with_warmup
import torch
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from torch import nn
from tqdm import tqdm
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, confusion_matrix
device = "cuda" if torch.cuda.is_available() else "cpu"
MODEL_CHECKPOINT = "SamLowe/roberta-base-go_emotions"
````

````python
# Tokenizer of RoBERTa, aligned with the emotional checkpoint
tokenizer = AutoTokenizer.from_pretrained(MODEL_CHECKPOINT, cache_dir="./cache/")

# additional special tokens
additional = ['[mention]', '[elip]', '[name]', '[religion]', '[team]']
tokenizer.add_special_tokens({'additional_special_tokens': additional})
````

````python
# Dataset class
class SentimentDataset(Dataset):
    def __init__(self, df, max_len=128, has_label=False, label_mapping=None):
        self.texts = df['final_input_text'].tolist()
        self.has_label = has_label
        self.max_len = max_len
        
        if self.has_label:
            self.labels = df['label'].tolist()

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = self.texts[idx]
        encoding = tokenizer.encode_plus(
            text,
            add_special_tokens=True,       
            max_length=self.max_len,
            truncation=True,               
            padding='max_length',          
            return_token_type_ids=True,   
            return_attention_mask=True,
            return_tensors='pt'            
        )
        
        item = {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
            'token_type_ids': encoding['token_type_ids'].flatten()
        }
        
        if self.has_label:
            item['labels'] = torch.tensor(self.labels[idx], dtype=torch.long)

        return item
````

````python
# define our collate_fn by overriding
class CustomDataCollator(DataCollatorWithPadding):
    def __call__(self, features):
        has_labels = 'labels' in features[0]
        if has_labels:
            label_name = "labels"
            labels = [feature.pop(label_name) for feature in features]
        
        batch = super().__call__(features)
        
        if has_labels:
            batch[label_name] = torch.stack(labels)
            
        return batch
    
custom_data_collator = CustomDataCollator(tokenizer=tokenizer)
````

````python
# train/val split with stratified
train, val = train_test_split(train_df, test_size=0.2, stratify=train_df['label'])

# max length
max_len = 128

# dataset 
train_dataset = SentimentDataset(train, has_label=True)
val_dataset = SentimentDataset(val, has_label=True)
test_dataset = SentimentDataset(test_df,has_label=False)

# dataloader
batch_size = 64
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, collate_fn=custom_data_collator)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=True, collate_fn=custom_data_collator)
test_loader = DataLoader(test_dataset, batch_size=1, shuffle=False, collate_fn=custom_data_collator)
````

````python
class SentimentRoBERTa(torch.nn.Module):
    def __init__(self, num_labels=6, num_layers_to_freeze=6):
        super().__init__()
        
        # Load the emotional-aware model's base encoder
        self.roberta = AutoModel.from_pretrained(MODEL_CHECKPOINT)
        
        # Freeze parameters
        # Freeze Embedding layer
        for param in self.roberta.embeddings.parameters():
            param.requires_grad = True

        # Freeze specified number of Transformer layers (0-5)
        for i, layer in enumerate(self.roberta.encoder.layer):
            if i < num_layers_to_freeze:
                for param in layer.parameters():
                    param.requires_grad = False
        
        # Classification Head
        hidden_size = self.roberta.config.hidden_size
        self.dropout = nn.Dropout(0.2)
        self.cls = nn.Linear(hidden_size, num_labels)

    def forward(self, input_ids, attention_mask=None):
        outputs = self.roberta(
            input_ids=input_ids,
            attention_mask=attention_mask,
        )
        
        # Get the hidden state of the [CLS] token (first token of the sequence)
        pooled_output = outputs.last_hidden_state[:, 0, :]
        pooled_output = self.dropout(pooled_output)

        cls_logits = self.cls(pooled_output)

        return cls_logits

# --- Model Instantiation and Setup ---

# Define constants
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
NUM_LABELS = 6
NUM_FREEZE_LAYERS = 0

# Instantiate model
model = SentimentRoBERTa(num_labels=NUM_LABELS, num_layers_to_freeze=NUM_FREEZE_LAYERS)

# Resize token embeddings to match the updated tokenizer vocabulary size
model.roberta.resize_token_embeddings(len(tokenizer))
model.to(DEVICE)
````

````python
# hyperparameters
LR_PEAK = 5e-6   # 峰值學習率 (您的成功值)
WARMUP_RATIO = 0.1
epochs = 15

num_training_steps = len(train_loader) * epochs
num_warmup_steps = int(WARMUP_RATIO * num_training_steps)

# optimizer
optimizer = AdamW(model.parameters(), lr=LR_PEAK)

# loss
class_counts = train['label'].value_counts().sort_index()
weights = 1.0 / np.sqrt(class_counts.values)
class_weights_tensor = torch.tensor(weights, dtype=torch.float).to(device)
cls_criterion = nn.CrossEntropyLoss(weight=class_weights_tensor)

# scheduler
scheduler = get_linear_schedule_with_warmup(
    optimizer, 
    num_warmup_steps=num_warmup_steps,
    num_training_steps=num_training_steps
)
````

````python
best_score = 0.0

for ep in range(epochs):
    pbar = tqdm(train_loader)
    pbar.set_description(f"Training epoch [{ep+1}/{epochs}]")
    model.train()
    
    # training
    n_batch = 0
    for batch in pbar:
        items = batch
        input_ids = items['input_ids'].to(device)
        attention_mask = items['attention_mask'].to(device)
        cls_labels = items['labels'].to(device).long()

        # clear gradient
        optimizer.zero_grad()

        # forward pass
        outputs = model(input_ids=input_ids, attention_mask=attention_mask) # 移除 token_type_ids

        cls_outputs = outputs

        # compute loss
        loss = cls_criterion(cls_outputs, cls_labels)

        # back-propagation
        loss.backward()

        # model optimization
        optimizer.step()
        
        # 加入 Scheduler Step
        scheduler.step()

        n_batch += 1

        pbar.set_postfix({
            "cls_loss": f"{loss.item():.4f}"
        })

    pbar = tqdm(val_loader)
    pbar.set_description(f"Validation epoch [{ep+1}/{epochs}]")
    model.eval()
    
    # evaluation
    all_cls_preds, all_cls_labels = [], []
    n_batch = 0

    with torch.no_grad():
        for batch in pbar:
            items = batch
            input_ids = items['input_ids'].to(device)
            attention_mask = items['attention_mask'].to(device)
            cls_labels = items['labels'].to(device).long()

            outputs = model(input_ids=input_ids, attention_mask=attention_mask) # 移除 token_type_ids
            cls_outputs = outputs

            preds = torch.argmax(cls_outputs, dim=1)
            all_cls_preds.extend(preds.cpu().numpy())
            all_cls_labels.extend(cls_labels.cpu().numpy())

            val_loss = cls_criterion(cls_outputs, cls_labels)
            pbar.set_postfix({"Val Loss": f"{val_loss.item():.4f}"})
            
    # Output all the evaluation scores 
    f1 = f1_score(all_cls_labels, all_cls_preds, average='macro')
    print(f"Evaluation Results - Epoch {ep+1}: Macro F1 = {f1:.4f}\n")
    
    if f1 > best_score:
        best_score = f1
        torch.save(model.state_dict(), f'./saved_models/best_model.ckpt')
````

````python
# fine-tuning with lower LR

# hyperparameters
LR = 1e-6   
epochs = 5

# optimizer
optimizer = AdamW(model.parameters(), lr=LR)
````

````python
best_score = 0.0

for ep in range(epochs):
    pbar = tqdm(train_loader)
    pbar.set_description(f"Training epoch [{ep+1}/{epochs}]")
    model.train()
    
    # training
    n_batch = 0
    for batch in pbar:
        items = batch
        input_ids = items['input_ids'].to(device)
        attention_mask = items['attention_mask'].to(device)
        cls_labels = items['labels'].to(device).long()

        # clear gradient
        optimizer.zero_grad()

        # forward pass
        outputs = model(input_ids=input_ids, attention_mask=attention_mask) # 移除 token_type_ids

        cls_outputs = outputs

        # compute loss
        loss = cls_criterion(cls_outputs, cls_labels)

        # back-propagation
        loss.backward()

        # model optimization
        optimizer.step()

        n_batch += 1

        pbar.set_postfix({
            "cls_loss": f"{loss.item():.4f}"
        })

    pbar = tqdm(val_loader)
    pbar.set_description(f"Validation epoch [{ep+1}/{epochs}]")
    model.eval()
    
    # evaluation
    all_cls_preds, all_cls_labels = [], []
    n_batch = 0

    with torch.no_grad():
        for batch in pbar:
            items = batch
            input_ids = items['input_ids'].to(device)
            attention_mask = items['attention_mask'].to(device)
            cls_labels = items['labels'].to(device).long()

            outputs = model(input_ids=input_ids, attention_mask=attention_mask) # 移除 token_type_ids
            cls_outputs = outputs

            preds = torch.argmax(cls_outputs, dim=1)
            all_cls_preds.extend(preds.cpu().numpy())
            all_cls_labels.extend(cls_labels.cpu().numpy())

            val_loss = cls_criterion(cls_outputs, cls_labels)
            pbar.set_postfix({"Val Loss": f"{val_loss.item():.4f}"})
            
    # Output all the evaluation scores 
    f1 = f1_score(all_cls_labels, all_cls_preds, average='macro')
    print(f"Evaluation Results - Epoch {ep+1}: Macro F1 = {f1:.4f}\n")
    
    if f1 > best_score:
        best_score = f1
        torch.save(model.state_dict(), f'./saved_models/best_model_ft.ckpt')
````
